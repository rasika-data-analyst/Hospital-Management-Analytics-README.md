```python
patients = pd.read_sql("SELECT * FROM patients_mysql", engine)
appointments = pd.read_sql("SELECT * FROM appointments", engine)
admissions = pd.read_sql("SELECT * FROM admissions", engine)
billing = pd.read_sql("SELECT * FROM billing", engine)
lab_tests = pd.read_sql("SELECT * FROM lab_tests", engine)
prescriptions = pd.read_sql("SELECT * FROM prescriptions", engine)
doctors = pd.read_sql("SELECT * FROM doctors", engine)
departments = pd.read_sql("SELECT * FROM departments", engine)
medicines = pd.read_sql("SELECT * FROM medicines", engine)
rooms = pd.read_sql("SELECT * FROM rooms", engine)
```

```python
tables = {
    "patients": patients,
    "appointments": appointments,
    "admissions": admissions,
    "billing": billing,
    "lab_tests": lab_tests,
    "prescriptions": prescriptions,
    "doctors": doctors,
    "departments": departments,
    "medicines": medicines,
    "rooms": rooms
}

for name, df in tables.items():
    print(f"{name:15} → {df.shape}")
```

```python
def profile_tables(tables):
    for name, df in tables.items():
        print("\n" + "=" * 70)
        print(f"TABLE: {name.upper()}")
        print("=" * 70)

        print(f"Rows: {df.shape[0]:,}")
        print(f"Columns: {df.shape[1]}")

        print("\nData Types:")
        print(df.dtypes)

        print("\nMissing Values:")
        missing = df.isnull().sum()
        missing = missing[missing > 0]

        if missing.empty:
            print("No missing values")
        else:
            print(missing)

        print("\nDuplicate Rows:")
        print(df.duplicated().sum())
```

### Question 2 — Department Performance

```python
department_performance = (
    appointments
    .merge(
        departments[["department_id", "department_name"]],
        on="department_id",
        how="left"
    )
    .groupby("department_name")
    .agg(
        total_appointments=("appointment_id", "nunique"),
        unique_patients=("patient_id", "nunique"),
        avg_wait_time=("wait_time_min", "mean")
    )
    .reset_index()
    .sort_values("total_appointments", ascending=False)
)

department_performance
```

### 2 — Patient & Appointment Behavior

```python
patient_behavior = appointments.merge(
    patients[["patient_id", "gender", "age", "city", "insurance_provider"]],
    on="patient_id",
    how="left"
)

patient_behavior.groupby("gender").agg(
    total_appointments=("appointment_id", "nunique"),
    unique_patients=("patient_id", "nunique"),
    avg_age=("age", "mean"),
    avg_satisfaction=("satisfaction_rating", "mean")
).reset_index()
```

#### Analysis 2 — Age Segmentation

```python
patient_behavior["age_group"] = pd.cut(
    patient_behavior["age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["0-18", "19-30", "31-45", "46-60", "60+"]
)

age_analysis = (
    patient_behavior
    .groupby("age_group", observed=True)
    .agg(
        appointments=("appointment_id", "nunique"),
        unique_patients=("patient_id", "nunique"),
        avg_satisfaction=("satisfaction_rating", "mean")
    )
    .reset_index()
)

age_analysis
```

```python
wait_satisfaction = appointments[
    ["appointment_id", "status", "wait_time_min", "satisfaction_rating"]
].dropna(
    subset=["wait_time_min", "satisfaction_rating"]
).copy()
```

```python
correlation = wait_satisfaction[
    ["wait_time_min", "satisfaction_rating"]
].corr().iloc[0, 1]

print("Correlation between Wait Time and Satisfaction:",
      round(correlation, 3))
```

```python
wait_satisfaction["wait_group"] = pd.cut(
    wait_satisfaction["wait_time_min"],
    bins=[-1, 15, 30, 45, 60, float("inf")],
    labels=["0-15 min", "16-30 min", "31-45 min", "46-60 min", "60+ min"]
)
```

```python
wait_analysis = (
    wait_satisfaction
    .groupby("wait_group", observed=True)
    .agg(
        appointments=("appointment_id", "nunique"),
        avg_satisfaction=("satisfaction_rating", "mean")
    )
    .reset_index()
)

wait_analysis
```

```python
q1 = billing_clean["total_amount"].quantile(0.25)
q3 = billing_clean["total_amount"].quantile(0.75)

iqr = q3 - q1

upper_limit = q3 + 1.5 * iqr

high_value_bills = billing_clean[
    billing_clean["total_amount"] > upper_limit
]

print("Q1:", round(q1, 2))
print("Q3:", round(q3, 2))
print("Upper Outlier Limit:", round(upper_limit, 2))
print("High-value bills:", len(high_value_bills))
print(
    "High-value revenue:",
    round(high_value_bills["total_amount"].sum(), 2)
)
print(
    "Share of total revenue:",
    round(
        high_value_bills["total_amount"].sum()
        / billing_clean["total_amount"].sum() * 100,
        2
    ),
    "%"
)
```

```python
reference_date = pd.Timestamp("2025-12-31")

calculated_age = (
    reference_date.year - patients["date_of_birth"].dt.year
    - ((patients["date_of_birth"].dt.month > reference_date.month)|
        ((patients["date_of_birth"].dt.month == reference_date.month) &
        (patients["date_of_birth"].dt.day > reference_date.day))))

age_mismatch = patients[
    patients["age"] != calculated_age
]

print("Age mismatches:", len(age_mismatch))
```

```python
no_show_rate = (
    (appointments["status"] == "No-Show").sum()/ len(appointments)* 100)

print("Total Appointments:", len(appointments))
print("No-Show Appointments:", (appointments["status"] == "No-Show").sum())
print("Overall No-Show Rate:", round(no_show_rate, 2), "%")
```

```python
no_show_by_type = (
    appointments.groupby("appointment_type")
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_type["no_show_rate"] = (
    no_show_by_type["no_shows"]
    / no_show_by_type["total_appointments"]
    * 100
)

display(no_show_by_type.sort_values("no_show_rate", ascending=False))
```

```python
no_show_by_dept = (
    appointments.groupby("department_id")
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_dept["no_show_rate"] = (
    no_show_by_dept["no_shows"]
    / no_show_by_dept["total_appointments"]
    * 100
)

display(
    no_show_by_dept
    .sort_values("no_show_rate", ascending=False)
)
```

```python
display(
    departments[
        ["department_id", "department_name"]
    ]
)
```

```python
no_show_by_dept = (
    no_show_by_dept
    .reset_index()
    .merge(
        departments[["department_id", "department_name"]],
        on="department_id",
        how="left"
    )
)

display(
    no_show_by_dept[
        ["department_id", "department_name",
         "total_appointments", "no_shows", "no_show_rate"]
    ]
    .sort_values("no_show_rate", ascending=False)
)
```

```python
no_show_by_gender = (
    appointments
    .merge(
        patients[["patient_id", "gender"]],
        on="patient_id",
        how="left"
    )
    .groupby("gender")
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_gender["no_show_rate"] = (
    no_show_by_gender["no_shows"]
    / no_show_by_gender["total_appointments"]
    * 100
)

display(no_show_by_gender)
```

```python
appointment_age = appointments.merge(
    patients[["patient_id", "age"]],
    on="patient_id",
    how="left"
)

appointment_age["age_group"] = pd.cut(
    appointment_age["age"],
    bins=[0, 18, 35, 50, 65, 100],
    labels=["0-18", "19-35", "36-50", "51-65", "66+"]
)

no_show_by_age = (
    appointment_age
    .groupby("age_group", observed=True)
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_age["no_show_rate"] = (
    no_show_by_age["no_shows"]
    / no_show_by_age["total_appointments"]
    * 100
)

display(no_show_by_age)
```

```python
appointments["appointment_date"] = pd.to_datetime(
    appointments["appointment_date"]
)

appointments["day_of_week"] = appointments["appointment_date"].dt.day_name()

no_show_by_day = (
    appointments
    .groupby("day_of_week")
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_day["no_show_rate"] = (
    no_show_by_day["no_shows"]
    / no_show_by_day["total_appointments"]
    * 100
)

day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

no_show_by_day = no_show_by_day.reindex(day_order)

display(no_show_by_day)
```

```python
appointments["wait_time_group"] = pd.cut(
    appointments["wait_time_min"],
    bins=[-1, 15, 30, 60, 120, float("inf")],
    labels=["0-15 min", "16-30 min", "31-60 min", "61-120 min", "120+ min"]
)

no_show_by_wait = (
    appointments
    .groupby("wait_time_group", observed=True)
    .agg(
        total_appointments=("appointment_id", "count"),
        no_shows=("status", lambda x: (x == "No-Show").sum())
    )
)

no_show_by_wait["no_show_rate"] = (
    no_show_by_wait["no_shows"]
    / no_show_by_wait["total_appointments"]
    * 100
)

display(no_show_by_wait)
```
